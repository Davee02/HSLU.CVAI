# Canny Edge Detector

```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as pl
import ipywidgets as iw
%matplotlib widget
```

## Bild einlesen

```python
img_rgb = cv.imread ('geraden.jpg')
print (img_rgb.shape)
img  = cv.cvtColor (img_rgb, cv.COLOR_BGR2GRAY)
print (img.shape)
pl.imshow (img, cmap = 'gray')
```

## Berechnung

```python
def CannyGui ():

    fig = pl.figure()
    fig.canvas.toolbar_visible = False
    fig.canvas.header_visible = False
    fig.canvas.footer_visible = False
    wdg = pl.imshow (img, cmap = 'gray')

    def Callback (tmin, tmax, aps, l2):
        edg = cv.Canny (img, tmin, tmax, apertureSize = aps, L2gradient = l2)
        wdg.set_data (edg)

    Slider1 = iw.IntSlider (min = 0, max = 255, step = 1, value = 128, description = 'Min. Threshold')
    Slider2 = iw.IntSlider (min = 0, max = 255, step = 1, value = 128, description = 'Max. Threshold')
    Dropdown = iw.Dropdown (options = [3, 5, 7], value = 3, description = 'Sobel Size', )
    Checkbox = iw.Checkbox (value = True, description = 'L2 Gradient', indent = False)
    UserInterface = iw.HBox ([Slider1, Slider2, Dropdown, Checkbox])
    display (UserInterface)
    Output = iw.interactive_output (Callback, {'tmin': Slider1, 'tmax': Slider2, 'aps': Dropdown, 'l2': Checkbox})

CannyGui()
```
